## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
print("There is no target. This is the key thing that separates clustering from the other lanes. I'm not defining a label at all.")
print("Instead of what am I predicting, the honest question is what am I grouping by. \nI'll group pages using observed performance signals - traffic volume, position, trend, engagement, freshness - chosen before clustering, not derived from any product rule.")
print("If I ever wanted to check whether clusters line up with something like future decline, that would be a validation step afterward, not a target I train on.")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
print("Silhouette score, backed by a human sense-check.")
print("Silhouette score gives me a number for how well-separated are these clusters (range -1 to 1, higher = more distinct groups) that's the defensible, computable half. ")
print("But a high silhouette score on meaningless clusters is still meaningless, so the real bar is: \ncan I look at 5-10 pages from each cluster and describe, in plain language, what makes that group distinct? \nIf I can't explain a cluster in one sentence, it doesn't count as good, no matter the score.")
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
print("One row = one content page")
import pandas as pd

df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

# Confirm grain: one row per content_id, no duplicates
print(df.shape)
print(df['content_id'].nunique(), "unique content items")
print(df[['content_id','client_id','impressions_90d','avg_position',
           'ctr','engagement_rate','trend_direction','content_age_days']].head())

print("\nExpected: df.shape[0] == df['content_id'].nunique() - that confirms the grain is truly one row per page, \nwhich matters before I run any clustering (duplicate rows would silently distort cluster centroids).")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
print("A fixed rule can only check one or two conditions at a time - e.g. if impressions > 10,000, call it a champion.")
print("But whether a page is a hidden gem, stale-visible, or declining champion depends on several signals interacting at once: \ntraffic level combined with trend direction combined with position combined with engagement.")
print("Writing nested if/else logic to capture every meaningful combination of 5+ continuous variables gets unmanageable fast, and any threshold I hand-pick is arbitrary.")
print("Clustering instead lets the data itself find where the natural groupings sit? which combinations of signals actually co-occur? rather than me guessing the boundaries in advance")
print("That matches the framing skill's rule exactly: ML earns its place when 'the pattern is real but too messy to write by hand.'")
```
